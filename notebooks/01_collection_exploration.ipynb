**1.5. Dòng trùng lặp và dòng trống**

Ở bước cuối cùng của phần Overview, nhóm sẽ kiểm tra số tên ứng dụng khác nhau, số dòng trùng lặp hoàn toàn và số dòng có tất cả các cột đều thiếu dữ liệu. Dòng chỉ thiếu một số thuộc tính được xem xét riêng trong các phần sau.

```python
# ---Thống kê số dòng và bản ghi trùng lặp---
overview = pd.Series({
    "Rows": len(df_raw),
    "Distinct app names": df_raw["App"].nunique(),
    "Exact duplicate rows": df_raw.duplicated().sum(),
    "Fully empty rows": df_raw.isna().all(axis=1).sum()
})
display(overview)
```

Bộ dữ liệu có **9.660 tên ứng dụng khác nhau**, **483 dòng trùng lặp hoàn toàn** và **0 dòng có tất cả các cột đều thiếu dữ liệu**.

Tiếp theo, nhóm kiểm tra một vài ví dụ về các dòng trùng lặp và kiểm tra các bản ghi có cùng tên ứng dụng.

```python
# ---Hiển thị nhóm bản ghi trùng hoàn toàn đầu tiên---
print("1. Exact duplicate rows:")
duplicated_rows = df_raw.loc[df_raw.duplicated(keep=False)]

if duplicated_rows.empty:
    print("No exact duplicate rows found.")
else:
    first_duplicate = duplicated_rows.iloc[[0]]
    display(df_raw.merge(first_duplicate, how="inner"))

# ---Hiển thị các bản ghi của tên ứng dụng bị lặp đầu tiên---
print("2. Repeated app names:")
duplicated_names = df_raw.loc[
    df_raw["App"].duplicated(keep=False), "App"
].dropna()

if duplicated_names.empty:
    print("No repeated app names found.")
else:
    first_name = duplicated_names.iloc[0]
    display(df_raw.loc[
        df_raw["App"] == first_name,
        ["App", "Category", "Rating", "Reviews"]
    ])
```

**Quyết định cho bước tiền xử lý sau này:**

- Dự kiến loại 483 dòng trùng lặp hoàn toàn, giữ lần xuất hiện đầu tiên. Các bản sao không bổ sung thông tin và có thể khiến một số bản ghi được tính nhiều lần trong các lần thống kê.
- Không có dòng trống hoàn toàn cần loại bỏ do kết quả kiểm tra cho thấy số dòng có tất cả các cột đều thiếu dữ liệu là 0. Các dòng chỉ thiếu một vài thuộc tính sẽ được xem xét và đánh giá trong các phần sau.
- Chưa loại các dòng chỉ trùng tên ứng dụng. Ví dụ như các bản ghi có thể có số lượng bài đánh giá khác nhau, hay điểm đánh giá thay đổi. Nhóm sẽ điều tra và phân tích thêm thêm trước khi chọn hoặc tổng hợp các bản ghi cùng tên.

```python
# ---Tạo bảng số tạm để kiểm tra---
numeric_view = pd.DataFrame(index=df_raw.index)
numeric_view["Rating"] = pd.to_numeric(df_raw["Rating"], errors="coerce")
numeric_view["Reviews"] = pd.to_numeric(df_raw["Reviews"], errors="coerce")

# ---Bỏ dấu phân cách và ký hiệu ngưỡng cài đặt---
installs_text = df_raw["Installs"].astype("string")
installs_text = installs_text.str.replace(",", "", regex=False)
installs_text = installs_text.str.replace("+", "", regex=False)
numeric_view["Installs"] = pd.to_numeric(installs_text, errors="coerce")

# ---Bỏ ký hiệu tiền tệ---
price_text = df_raw["Price"].astype("string")
price_text = price_text.str.replace("$", "", regex=False)
numeric_view["Price"] = pd.to_numeric(price_text, errors="coerce")

# ---Trích xuất giá trị và đơn vị kích thước---
size_text = df_raw["Size"].astype("string")
size_parts = size_text.str.extract(r"^(\d+(?:\.\d+)?)([Mk])$")
size_number = pd.to_numeric(size_parts[0], errors="coerce")
size_multiplier = size_parts[1].map({"M": 1.0, "k": 1 / 1024})
numeric_view["Size"] = size_number * size_multiplier
numeric_view = numeric_view[numerical_cols]
```

Sau khi tạo bản ghi tạm, nhóm sẽ cho kiểm tra bản ghi này để xem bộ dữ liệu có chứa giá trị thiếu, placeholder và lỗi chuyển đổi hay không

```python
# ---Phân biệt giá trị thiếu, nhãn thay thế và lỗi chuyển đổi---
quality_records = []

for col in numerical_cols:
    text = df_raw[col].astype("string")
    original_missing = text.isna() | (text.str.strip() == "").fillna(False)

    placeholder = pd.Series(False, index=df_raw.index)
    if col == "Size":
        placeholder = (text == "Varies with device").fillna(False)

    parse_failed = numeric_view[col].isna() & ~original_missing & ~placeholder
    quality_records.append({
        "Column": col,
        "Original missing": int(original_missing.sum()),
        "Placeholder": int(placeholder.sum()),
        "Parse failed": int(parse_failed.sum())
    })

    if parse_failed.any():
        print(f"{col}: values that could not be parsed")
        display(df_raw.loc[parse_failed, ["App", col]])

quality_summary = pd.DataFrame(quality_records).set_index("Column")
display(quality_summary)
```

```python
# ---Kiểm tra miền hợp lệ của các giá trị số---
invalid_masks = {}
invalid_masks["Rating"] = (
    (numeric_view["Rating"] < 1) | (numeric_view["Rating"] > 5)
)

for col in ["Reviews", "Installs"]:
    values = numeric_view[col]
    invalid_masks[col] = values.notna() & ((values < 0) | (values % 1 != 0))

invalid_masks["Size"] = numeric_view["Size"] <= 0
invalid_masks["Price"] = numeric_view["Price"] < 0

for col in numerical_cols:
    mask = invalid_masks[col]
    print(f"{col}: {mask.sum()} invalid numerical values")
    display(df_raw.loc[mask, ["App", col]])
```

**2.4. Giá trị ngoại lệ và giá trị bất thường**  

Ở bước cuối nhóm sẽ tìm các điểm cực trị, đối chiếu bản ghi và phân biệt lỗi dữ liệu với giá trị lớn nhưng hợp lệ.

```python
# ---Đánh dấu các giá trị nằm ngoài hàng rào IQR---
outlier_masks = {}
outlier_records = []

for col in numerical_cols:
    values = numeric_stats_view[col]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    mask = values.notna() & ((values < lower) | (values > upper))
    outlier_masks[col] = mask

    valid_count = values.notna().sum()
    percent = np.nan
    if valid_count > 0:
        percent = mask.sum() / valid_count * 100

    outlier_records.append({
        "Column": col, "Lower bound": lower, "Upper bound": upper,
        "IQR": iqr, "Flagged count": int(mask.sum()), "Flagged percent": percent
    })

outlier_summary = pd.DataFrame(outlier_records).set_index("Column")
display(outlier_summary.round(3))
```

**Nhận xét về kết quả sàng lọc bằng IQR:**

- `Rating` có **503 bản ghi (5,37%)** nằm ngoài hàng rào IQR. Do cận dưới là **3,25** và cận trên là **5,25**, các bản ghi được đánh dấu đều thuộc phía điểm thấp. Điểm thấp vẫn có thể hợp lệ nếu nằm trong thang 1–5.
- `Reviews` có **1.924 bản ghi (17,75%)**, `Size` có **564 bản ghi (6,17%)** và `Installs` có **828 bản ghi (7,64%)** được đánh dấu. Các biến này không âm và có cận dưới IQR âm, nên những giá trị được đánh dấu nằm phía trên cận trên, lần lượt là **136.881,75 lượt đánh giá**, **67,65 MB** và **12.498.500 lượt cài đặt theo cận dưới**.
- `Price` có **IQR bằng 0**, khiến cả hai hàng rào đều bằng 0. Vì vậy, toàn bộ **800 bản ghi có giá dương (7,38%)** đều được đánh dấu. Quy tắc này không phân biệt được mức giá trả phí thông thường với giá cao cần kiểm tra thêm.


```python
# ---Xem năm giá trị thấp nhất và cao nhất được đánh dấu---
for col in numerical_cols:
    flagged_values = numeric_stats_view.loc[outlier_masks[col], col]
    low_values = flagged_values.sort_values(kind="stable").head(5)
    high_values = flagged_values.sort_values(ascending=False, kind="stable").head(5)
    selected_values = pd.concat([low_values, high_values])
    selected_index = selected_values.index.unique()

    examples = df_raw.loc[selected_index, ["App", "Category", col]].copy()
    examples["Numeric value"] = numeric_stats_view.loc[selected_index, col]
    print(f"{col}: examples flagged by the IQR rule")
    display(examples.sort_values("Numeric value"))
```

**Nhận xét khi đối chiếu các bản ghi được đánh dấu:**

- Những ví dụ của `Rating` có điểm từ **1,0 đến 3,2**, vẫn nằm trong thang đánh giá hợp lệ. Chưa có bằng chứng từ các bản ghi được hiển thị để xem điểm thấp là lỗi nhập liệu.
- Các giá trị lớn của `Reviews` xuất hiện ở những tên như `Facebook` và `WhatsApp Messenger`; các ngưỡng `Installs` cao xuất hiện ở `Gmail`, `Google Chrome: Fast & Secure` và các ứng dụng khác. Những ví dụ này phù hợp với khả năng dữ liệu có các ứng dụng rất phổ biến, nhưng tên ứng dụng tự nó chưa đủ để xác minh độ chính xác của số liệu.
- Các ví dụ `Size` có kích thước từ **68 đến 100 MB** và vẫn có đơn vị hợp lệ. Không có căn cứ kết luận đây là lỗi chỉ vì kích thước lớn hơn phần lớn bản ghi.
- Các ví dụ `Price` trải từ **0,99 đến 400 USD**, cho thấy quy tắc IQR đánh dấu cả mức giá thấp thông thường lẫn giá rất cao. Nhóm cần xem xét riêng các mức giá cao thay vì loại toàn bộ các giá trị được đánh dấu.
- Cell chỉ hiển thị tối đa năm giá trị nhỏ nhất và năm giá trị lớn nhất trong nhóm được đánh dấu của mỗi cột, không phải toàn bộ ngoại lệ. Một số tên ứng dụng xuất hiện nhiều lần vì dữ liệu vẫn giữ các bản ghi trùng lặp trong giai đoạn khám phá.


```python
# ---Tạo bảng tạm để kiểm tra các bản ghi---
base = df_raw.loc[numeric_stats_view.index].copy()
base["Rating_num"] = numeric_stats_view["Rating"]
base["Reviews_num"] = numeric_stats_view["Reviews"]
base["Installs_num"] = numeric_stats_view["Installs"]
base["Price_num"] = numeric_stats_view["Price"]
base["Size_MB"] = numeric_stats_view["Size"]
base["Last_Updated_dt"] = pd.to_datetime(base["Last Updated"], errors="coerce")

# ---Kiểm tra độ lệch và ngoại lệ theo điểm chuẩn hóa---
z_rows = []
for col in numerical_cols:
    values = numeric_stats_view[col].dropna()
    z_scores = (values - values.mean()) / values.std(ddof=1)
    flagged = np.abs(z_scores) > 3
    z_rows.append({
        "Column": col,
        "Skew": values.skew(),
        "Absolute z > 3 count": int(flagged.sum()),
        "Absolute z > 3 percent": flagged.mean() * 100
    })
display(pd.DataFrame(z_rows).round(3))

# ---Khai báo ngưỡng cài đặt kế tiếp---
install_upper_limits = {
    0: 1, 1: 5, 5: 10, 10: 50, 50: 100, 100: 500, 500: 1_000,
    1_000: 5_000, 5_000: 10_000, 10_000: 50_000, 50_000: 100_000,
    100_000: 500_000, 500_000: 1_000_000, 1_000_000: 5_000_000,
    5_000_000: 10_000_000, 10_000_000: 50_000_000,
    50_000_000: 100_000_000, 100_000_000: 500_000_000,
    500_000_000: 1_000_000_000, 1_000_000_000: np.inf
}

# ---Đối chiếu số lượt đánh giá với ngưỡng cài đặt---
base["Install_upper"] = base["Installs_num"].map(install_upper_limits)
above_floor = base["Reviews_num"] > base["Installs_num"]
reach_next = base["Reviews_num"] >= base["Install_upper"]
dedup = base.drop_duplicates(subset=df_raw.columns)
category_counts = dedup.groupby("App")["Category"].nunique()
repeated_apps = dedup.loc[dedup["App"].duplicated(keep=False), "App"]

record_checks = pd.Series({
    "Last Updated min": str(base["Last_Updated_dt"].min().date()),
    "Last Updated max": str(base["Last_Updated_dt"].max().date()),
    "Rating below 3.25": int((base["Rating"] < 3.25).sum()),
    "Rating above 5.25": int((base["Rating"] > 5.25).sum()),
    "Size_MB equal to 100": int((base["Size_MB"] == 100).sum()),
    "Price >= 100": int((base["Price_num"] >= 100).sum()),
    "Reviews above install floor": int(above_floor.sum()),
    "Reviews at next install bucket": int(reach_next.sum()),
    "Install label 0 or 0+": int(base["Installs"].isin(["0", "0+"]).sum()),
    "Reviews mean, with exact duplicates": round(base["Reviews_num"].mean(), 1),
    "Reviews mean, exact duplicates removed": round(dedup["Reviews_num"].mean(), 1),
    "Extra rows after exact dedup": int(dedup["App"].duplicated().sum()),
    "Apps that still have extra rows": repeated_apps.nunique(),
    "Apps with more than one category": int((category_counts > 1).sum())
})
display(record_checks)
display(base.loc[base["Price_num"] >= 100,
                 ["App", "Category", "Price", "Installs", "Rating"]])
display(base.loc[above_floor,
                 ["App", "Reviews_num", "Installs", "Installs_num", "Install_upper"]])
```

**Nhận xét từ kiểm tra bổ sung:**

- Hệ số lệch của `Rating` là **−1,85**, còn `Reviews`, `Size`, `Installs` và `Price` đều dương. `Reviews` và `Price` có độ lệch lớn, lần lượt khoảng **16,45** và **23,71**. Kết quả phù hợp với hình dạng phân bố đã quan sát ở mục 2.3.
- Với điều kiện **|z| > 3**, số bản ghi được đánh dấu lần lượt là **175** ở `Rating`, **106** ở `Reviews`, **211** ở `Size`, **130** ở `Installs` và **24** ở `Price`.
```python
# ---Kiểm tra giá trị thiếu, chuỗi rỗng và khoảng trắng---
missing_mask = df_raw.isna()
for col in df_raw.columns:
    text = df_raw[col].astype("string")
    blank = (text.str.strip() == "").fillna(False)
    missing_mask[col] = missing_mask[col] | blank

missing_summary = pd.DataFrame({
    "Missing count": missing_mask.sum(),
    "Missing percent": missing_mask.mean() * 100
})
missing_summary = missing_summary.sort_values("Missing count", ascending=False)
display(missing_summary.round(2))
```

**Nhận xét:**  

- Cột `Rating` là cột thiếu nhiều giá trị nhất, với 1474 giá trị thiếu, chiếm 13.6% tổng số bản ghi.  
- Các cột khác như `Current Ver`, `Android Ver`, `Type`, `Content Rating` chỉ thiếu một vài giá trị (từ 1-8 giá trị), chiếm tỉ lệ rất nhỏ (chỉ từ 0.01% - 0.07%)
- Các cột còn lại không ghi nhận giá trị thiếu theo phép kiểm tra hiện tại.
- Cột `Size` không thiếu theo phép kiểm tra này. Nhãn `Varies with device` là placeholder, đã được tách ở mục 2.1, và không được cộng vào số giá trị thiếu.

**4.2. Mẫu xuất hiện của dữ liệu thiếu:**

Nhóm kiểm tra các tổ hợp cột bị thiếu trong cùng một bản ghi để xác định dữ liệu thường thiếu riêng lẻ hay đồng thời ở nhiều thuộc tính.


```python
# ---Liệt kê các cột bị thiếu trong từng bản ghi---
missing_patterns = []
for flags in missing_mask.values:
    missing_columns = []
    for col, is_missing in zip(df_raw.columns, flags):
        if is_missing:
            missing_columns.append(col)

    if missing_columns:
        missing_patterns.append(", ".join(missing_columns))
    else:
        missing_patterns.append("No missing values")

# ---Đếm số lần xuất hiện của từng mẫu dữ liệu thiếu---
pattern_counts = pd.Series(missing_patterns).value_counts()
pattern_summary = pd.DataFrame({
    "Count": pattern_counts,
    "Percent": pattern_counts / len(df_raw) * 100
})
display(pattern_summary.round(2))
```

**Nhận xét:**

Trường hợp cuối thuộc bản ghi có dấu hiệu lệch cột đã xác định ở mục 3.3, nên cần phân biệt với các trường hợp thiếu thông thường.

**5.5. Kích thước thay đổi tùy thiết bị và mức độ phổ biến**

Ở mục 2.1, nhãn `Varies with device` chỉ được thống kê tần suất. Tại đây, nhóm đối chiếu nhóm nhãn này với nhóm có kích thước số xác định về ngưỡng cài đặt, điểm đánh giá và mô hình thu phí. Biến đánh dấu chỉ phục vụ bảng kiểm tra tạm; không điền kích thước chưa xác định hoặc thay đổi `df_raw`.

```python
# ---Đối chiếu các nhóm kích thước trong bảng tạm---
size_relationship = paired[["App", "Category", "Type"]].copy()
size_relationship["Size group"] = "Fixed size"
size_varies = paired["Size"] == "Varies with device"
size_relationship.loc[size_varies, "Size group"] = "Varies with device"
size_relationship["Rating_num"] = numeric_stats_view["Rating"]
size_relationship["Installs_num"] = numeric_stats_view["Installs"]
size_relationship["Reached one million"] = size_relationship["Installs_num"] >= 1_000_000
size_relationship["Missing rating"] = size_relationship["Rating_num"].isna()

size_relationship_summary = size_relationship.groupby("Size group").agg(
    Records=("App", "size"),
    Median_install_floor=("Installs_num", "median"),
    Reached_one_million_percent=("Reached one million", "mean"),
    Missing_rating_percent=("Missing rating", "mean"),
    Median_rating=("Rating_num", "median")
)
size_relationship_summary["Reached_one_million_percent"] *= 100
size_relationship_summary["Missing_rating_percent"] *= 100
display(size_relationship_summary.round(2))

# ---Đếm từng loại ứng dụng trong mỗi nhóm kích thước---
size_relationship["Type group"] = size_relationship["Type"].astype("string").fillna("Missing")
size_type_counts = size_relationship.groupby(["Size group", "Type group"]).size().unstack(fill_value=0)
print("Type counts by size group:")
display(size_type_counts)

# ---Tính tỷ lệ loại ứng dụng trong mỗi nhóm kích thước---
size_type_percentages = size_type_counts.copy()
size_totals = size_type_counts.sum(axis=1)
for col in size_type_counts.columns:
    size_type_percentages[col] = size_type_counts[col] / size_totals * 100
print("Type percentages by size group:")
display(size_type_percentages.round(2))
```

**5.6. Độ phủ của bảng bình luận theo danh mục ứng dụng**

Câu hỏi 2.2 sử dụng bảng bình luận, nên nhóm kiểm tra mức độ khớp theo tên và độ phủ giữa các danh mục trước khi phân tích. Một ứng dụng được đánh dấu có bình luận nếu tên xuất hiện ở ít nhất một dòng có nội dung không rỗng. Nhóm chưa tổng hợp cảm xúc hoặc trả lời câu hỏi chuyên sâu tại đây.

Bảng tạm ứng dụng–danh mục chỉ giữ mỗi cặp tên và danh mục một lần để đếm độ phủ; dữ liệu gốc không bị thay đổi. Các tỷ lệ là độ phủ theo tên, không khẳng định tên ứng dụng là mã định danh duy nhất.

```python
# ---Kiểm tra nội dung bình luận và đối chiếu tên ứng dụng---
review_text = reviews_raw["Translated_Review"].astype("string")
review_has_text = review_text.notna() & (review_text.str.strip() != "").fillna(False)
review_names_with_text = set(reviews_raw.loc[review_has_text, "App"].dropna())
app_names_for_coverage = set(paired["App"].dropna())

review_coverage_overview = pd.Series({
    "Raw review rows": len(reviews_raw),
    "Rows with review text": int(review_has_text.sum()),
    "Rows with missing or empty text": int((~review_has_text).sum()),
    "Exact duplicate review rows": int(reviews_raw.duplicated().sum()),
    "App names with review text": len(review_names_with_text),
    "Names matched to the app table": len(review_names_with_text & app_names_for_coverage),
    "Names absent from the app table": len(review_names_with_text - app_names_for_coverage)
})
display(review_coverage_overview)

# ---Đếm độ phủ bình luận theo từng cặp ứng dụng và danh mục---
app_review_coverage = paired[["App", "Category"]].drop_duplicates().copy()
app_review_coverage["Has review text"] = app_review_coverage["App"].isin(review_names_with_text)
review_coverage_by_category = app_review_coverage.groupby("Category").agg(
    App_names=("App", "size"),
    Names_with_reviews=("Has review text", "sum"),
    Review_coverage_percent=("Has review text", "mean")
)
review_coverage_by_category["Review_coverage_percent"] *= 100
review_coverage_by_category = review_coverage_by_category.sort_values("Review_coverage_percent", ascending=False)
display(review_coverage_by_category.round(2))
```

**Nhận xét:**
