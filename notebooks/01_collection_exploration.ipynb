**3.1. Số lượng các giá trị phân biệt và các giá trị xuất hiện thường xuyên nhất**  

Ở bước này, nhóm tiếp tục tìm hiểu bốn cột mang ý nghĩa phân loại: `Category`, `Type`, `Content Rating` và `Genres`. Các cột này hiện được pandas đọc dưới dạng `object`, nhưng giá trị của chúng biểu diễn các nhãn hoặc nhóm, nên được xem xét trong phần dữ liệu phân loại.

Đầu tiên là nhóm sẽ trích xuất số lượng nhãn phân biệt (không tính giá trị NaN) và lấy ra nhãn xuất hiện thường xuyên nhất trong 4 cột phân loại:  

```python
categorical_cols = ["Category", "Type", "Content Rating", "Genres"]

frequency_tables = {}

for col in categorical_cols:
    counts = df_raw[col].value_counts(dropna=False)

    frequency_tables[col] = pd.DataFrame({
        "Count": counts,
        "Percent": counts / len(df_raw) * 100
    })

summary = pd.DataFrame({
    "Distinct labels": df_raw[categorical_cols].nunique(),
    "Most frequent label": [
        frequency_tables[c].index[0] for c in categorical_cols
    ],
    "Top-label percent": [
        frequency_tables[c]["Percent"].iloc[0]
        for c in categorical_cols
    ]
})

display(summary.round(2))
```

**Nhận xét:**  

- `Genres` có nhiều nhãn riêng biệt nhất với 120 nhãn, tiếp theo là `Category` với 34 nhãn, `Content Rating` có 6 nhãn và `Type` có 3 nhãn.
- "FAMILY", "Free", "Everyone" và "Tool" là số nhãn xuất hiện nhiều nhất trong 4 cột `Category`, `Type`, `Content Rating`, `Genres` (lần lượt là 18.19%, 92.6%, 80.38%, 7.77%) 

Trong cột `Genres`, số lượng nhãn riêng biệt lên tới 120, trong khi Google App Store chỉ hỗ trợ từ 30-50 genres khác nhau. Điều này là do mỗi tổ hợp thể loại được xem là một nhãn riêng biệt, ví dụ "Art & Design;Pretend Play" và "Art & Design" là hai nhãn riêng biệt

**3.4.  Danh mục nhãn hiếm cần được nhóm**  

Nhóm xác định nhãn hiếm là những nhãn xuất hiện trong ít hơn 20 bản ghi. Đây là ngưỡng sàng lọc do nhóm lựa chọn nhằm nhận diện các nhóm có ít dữ liệu để so sánh.  

Để tránh đưa các giá trị không hợp lệ vào danh sách nhãn hiếm, nhóm tạm thời loại bản ghi có dấu hiệu lệch cột khỏi phạm vi thống kê của mục này.

```python
# Temporary view excluding malformed records.
inspection_df = df_raw.loc[~suspect_rows]
rare_threshold = 20

for col in categorical_cols:
    counts = inspection_df[col].value_counts(dropna=False)

    table = pd.DataFrame({
        "Count": counts,
        "Percent": counts / len(inspection_df) * 100
    })

    rare_labels = table.loc[
        table.index.notna()
        & table["Count"].lt(rare_threshold)
    ]

    print(f"\n{col}: {len(rare_labels)} labels with fewer than {rare_threshold} records")

    with pd.option_context("display.max_rows", None):
        display(rare_labels.round(2))
```

**Nhận xét:**   

- Trong `Category` và `Type`, nhóm không nhận thấy nhãn hiếm. Tất cả nhãn đều xuất hiện nhiều hơn 20 lần.  
- Trong `Content Rating`, hai nhãn "Adults only 18+" và "Unrated" có ít bản ghi nhưng mang ý nghĩa khác nhau, nên nhóm chưa đề xuất gộp chung.  
- Cột `Genres` có 65 nhãn hiếm, chủ yếu là các tổ hợp thể loại. Nhóm sẽ cân nhắc tách các tổ hợp thành từng thành phần. Sau khi tách và đếm, nếu số lần nhãn riêng lẻ xuất hiện vẫn còn ít, nhóm sẽ cân nhắc gộp chúng lại vào nhãn "Other".
