**1.6. Bảng review**

Mỗi dòng của `googleplaystore_user_reviews.csv` là một review đã được dịch, không phải một ứng dụng. Bảng này thuộc cùng dataset trên Kaggle với bảng ứng dụng.

```python
reviews_raw = pd.read_csv(DATA_DIR / "googleplaystore_user_reviews.csv")

text_ok = reviews_raw["Translated_Review"].notna()
sentiment_ok = reviews_raw["Sentiment"].notna()
app_names = set(df_raw["App"])
review_names = set(reviews_raw["App"])
text_names = set(reviews_raw.loc[text_ok, "App"])

reviews_overview = pd.Series({
    "Rows": len(reviews_raw),
    "Columns": reviews_raw.shape[1],
    "Missing review text": int((~text_ok).sum()),
    "Rows with review text": int(text_ok.sum()),
    "Missing sentiment": int((~sentiment_ok).sum()),
    "Sentiment present but text missing": int((~text_ok & sentiment_ok).sum()),
    "Exact duplicate rows": int(reviews_raw.duplicated().sum()),
    "App names in reviews": len(review_names),
    "Review names found in app table": len(review_names & app_names),
    "Review names absent from app table": len(review_names - app_names),
    "Names with text and a row in the app table": len(text_names & app_names),
})

display(reviews_overview)
display(reviews_raw["Sentiment"].value_counts(dropna=False))
```

**Nhận xét:**

- Bảng review có **64.295 dòng** và **5 cột**: `App`, `Translated_Review`, `Sentiment`, `Sentiment_Polarity`, `Sentiment_Subjectivity`.
- **26.868** dòng thiếu nội dung review. **37.427** dòng còn nội dung. **26.863** dòng thiếu luôn Sentiment. **5** dòng không có nội dung nhưng vẫn có Sentiment.
- **33.616** dòng là bản sao hoàn toàn của một dòng đã xuất hiện trước đó.
- Bảng có **1.074** tên ứng dụng. **1.020** tên có trong bảng ứng dụng và **54** tên không có. **816** tên vừa có nội dung review vừa có trong bảng ứng dụng.
- Có 816 tên ứng dụng khớp có ít nhất một review chứa nội dung. Đây là phạm vi mẫu nếu phân tích yêu cầu nội dung review, không phải giới hạn của mọi kiểu gộp. Khi gộp cần tổng hợp review theo ứng dụng hoặc kiểm soát quan hệ nhiều–nhiều để tránh nhân bản bản ghi.

Nhóm sẽ xem xét năm cột biểu diễn số gồm `Rating`, `Reviews`, `Size`, `Installs` và `Price`. Một số cột như `Size` và `Installs` đang được lưu dưới dạng chuỗi do chứa đơn vị hoặc ký hiệu. Nhóm kiểm tra cách biểu diễn và tạo bảng số tạm để phục vụ thống kê, đồng thời phân biệt giá trị thiếu, giá trị chưa xác định và giá trị không hợp lệ.  

Trước hết, nhóm sẽ hiển thị trước một vài giá trị:

```python
numerical_cols = ["Rating", "Reviews", "Size", "Installs", "Price"]

display(df_raw[numerical_cols].head(5))
```

Trong năm bản ghi đầu, `Rating` được biểu diễn bằng số thập phân, còn `Reviews` và `Price` chứa số nguyên. Cột `Size` chứa hậu tố M, trong khi `Installs` chứa dấu phẩy và dấu +, nên cần chuyển đổi cách biểu diễn trước khi thống kê.

Tiếp theo nhóm sẽ tạo bảng số tạm để chuyển đổi cột `Size` và `Installs`. Với cột `Size`, nhóm đổi hậu tố `k` sang MB bằng cách chia cho 1.024, để giá trị `1020k` vẫn nhỏ hơn một ứng dụng ghi `1M`. Với các giá trị không chuyển được được biểu diễn bằng NaN trong bảng tạm để kiểm tra tiếp.

```python
numeric_view = pd.DataFrame(index=df_raw.index)

numeric_view["Rating"] = pd.to_numeric(
    df_raw["Rating"], errors="coerce"
)

numeric_view["Reviews"] = pd.to_numeric(
    df_raw["Reviews"], errors="coerce"
)

installs_text = (
    df_raw["Installs"].astype("string")
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)
numeric_view["Installs"] = pd.to_numeric(
    installs_text, errors="coerce"
)

price_text = (
    df_raw["Price"].astype("string")
    .str.replace("$", "", regex=False)
)
numeric_view["Price"] = pd.to_numeric(
    price_text, errors="coerce"
)

size_text = df_raw["Size"].astype("string")
size_parts = size_text.str.extract(r"^(\d+(?:\.\d+)?)([Mk])$")

size_number = pd.to_numeric(size_parts[0], errors="coerce")
size_multiplier = size_parts[1].map({"M": 1.0, "k": 1 / 1024})

numeric_view["Size"] = size_number * size_multiplier

numeric_view = numeric_view[numerical_cols]
```

Sau khi tạo bản ghi tạm, nhóm sẽ cho kiểm tra bản ghi này để xem bộ dữ liệu có chứa giá trị thiếu, placeholder và lỗi chuyển đổi hay không

```python
quality_records = []

for col in numerical_cols:
    raw_text = df_raw[col].astype("string")

    original_missing = (
        raw_text.isna()
        | raw_text.str.strip().eq("").fillna(False)
    )

    placeholder = pd.Series(False, index=df_raw.index)
    if col == "Size":
        placeholder = raw_text.eq("Varies with device").fillna(False)

    parse_failed = (
        numeric_view[col].isna()
        & ~original_missing
        & ~placeholder
    )

    quality_records.append({
        "Column": col,
        "Original missing": int(original_missing.sum()),
        "Placeholder": int(placeholder.sum()),
        "Parse failed": int(parse_failed.sum())
    })

    if parse_failed.any():
        print(f"\n{col}: values that could not be parsed")
        display(
            df_raw.loc[parse_failed, ["App", col]]
        )

display(pd.DataFrame(quality_records).set_index("Column"))
```

thể. Đây là placeholder, nhóm sẽ không xếp giá trị này thành giá trị thiếu hay giá trị có lỗi chuyển đổi  

- Các cột `Reviews`, `Size`, `Installs` và `Price` đều có một giá trị không chuyển được theo quy tắc đã đặt, lần lượt là 3.0M, 1,000+, Free và Everyone. Các giá trị này cùng thuộc bản ghi có index 10472, ứng dụng Life Made WI-Fi Touchscreen Photo Frame, cho thấy dấu hiệu lệch cột

Cuối cùng, nhóm sẽ cho kiểm tra tính hợp lệ của các giá trị số trong 5 cột này:  

```python
invalid_masks = {
    "Rating": (
        numeric_view["Rating"].notna()
        & ~numeric_view["Rating"].between(1, 5)
    ),
    "Reviews": (
        numeric_view["Reviews"].lt(0)
        | numeric_view["Reviews"].mod(1).ne(0)
    ) & numeric_view["Reviews"].notna(),
    "Size": numeric_view["Size"].le(0),
    "Installs": (
        numeric_view["Installs"].lt(0)
        | numeric_view["Installs"].mod(1).ne(0)
    ) & numeric_view["Installs"].notna(),
    "Price": numeric_view["Price"].lt(0)
}

for col, mask in invalid_masks.items():
    print(f"\n{col}: {mask.sum()} invalid numeric values")
    display(df_raw.loc[mask, ["App", col]])
```

**2.4. Giá trị ngoại lệ và giá trị bất thường**  

Ở bước cuối nhóm sẽ tìm các điểm cực trị, đối chiếu bản ghi và phân biệt lỗi dữ liệu với giá trị lớn nhưng hợp lệ.

```python
outlier_masks = {}
outlier_records = []

for col in numerical_cols:
    values = numeric_stats_view[col]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask = values.notna() & (
        values.lt(lower) | values.gt(upper)
    )
    outlier_masks[col] = mask

    valid_count = values.notna().sum()

    outlier_records.append({
        "Column": col,
        "Lower bound": lower,
        "Upper bound": upper,
        "IQR": iqr,
        "Flagged count": int(mask.sum()),
        "Flagged percent": (
            mask.sum() / valid_count * 100
            if valid_count else float("nan")
        )
    })

outlier_summary = pd.DataFrame(outlier_records).set_index("Column")
display(outlier_summary.round(3))
```

```python
for col in numerical_cols:
    flagged_values = numeric_stats_view.loc[outlier_masks[col], col]

    selected_index = (
        flagged_values.nsmallest(5).index
        .union(flagged_values.nlargest(5).index, sort=False)
    )

    examples = df_raw.loc[
        selected_index,
        ["App", "Category", col]
    ].copy()

    examples["Numeric value"] = numeric_stats_view.loc[
        selected_index, col
    ]

    print(f"\n{col}: examples flagged by the IQR rule")
    display(examples.sort_values("Numeric value"))
```

```python
import numpy as np

base = df_raw.loc[numeric_stats_view.index].copy()
base["Reviews_num"] = numeric_stats_view["Reviews"]
base["Installs_num"] = numeric_stats_view["Installs"]
base["Price_num"] = numeric_stats_view["Price"]
base["Size_MB"] = numeric_stats_view["Size"]
base["Last_Updated_dt"] = pd.to_datetime(base["Last Updated"], errors="coerce")

num_cols = ["Rating", "Reviews", "Size", "Installs", "Price"]
skew = numeric_stats_view[num_cols].skew()

z_rows = []
for col in num_cols:
    series = numeric_stats_view[col].dropna()
    z_hit = np.abs((series - series.mean()) / series.std(ddof=1)) > 3
    z_rows.append({
        "Column": col,
        "Skew": skew[col],
        "Z>3 outliers": int(z_hit.sum()),
        "Z>3 percent": z_hit.mean() * 100,
    })
display(pd.DataFrame(z_rows).round(3))

INSTALL_UPPER = {
    0: 1, 1: 5, 5: 10, 10: 50, 50: 100, 100: 500, 500: 1_000,
    1_000: 5_000, 5_000: 10_000, 10_000: 50_000, 50_000: 100_000,
    100_000: 500_000, 500_000: 1_000_000, 1_000_000: 5_000_000,
    5_000_000: 10_000_000, 10_000_000: 50_000_000, 50_000_000: 100_000_000,
    100_000_000: 500_000_000, 500_000_000: 1_000_000_000, 1_000_000_000: np.inf,
}

def install_upper(label):
    if label == "0":
        return 1.0
    floor = float(str(label).replace(",", "").replace("+", ""))
    return INSTALL_UPPER[floor]

base["Install_upper"] = base["Installs"].map(install_upper)
above_floor = base["Reviews_num"] > base["Installs_num"]
reach_next = base["Reviews_num"] >= base["Install_upper"]
dedup = base.drop_duplicates(subset=df_raw.columns)
display(pd.Series({
    "Last Updated min": base["Last_Updated_dt"].min().date().isoformat(),
    "Last Updated max": base["Last_Updated_dt"].max().date().isoformat(),
    "Rating below 3.25": int((base["Rating"].dropna() < 3.25).sum()),
    "Rating above 5.25": int((base["Rating"].dropna() > 5.25).sum()),
    "Size_MB equal to 100": int((base["Size_MB"] == 100).sum()),
    "Price >= 100": int((base["Price_num"] >= 100).sum()),
    "Reviews above install floor": int(above_floor.sum()),
    "Reviews at next install bucket": int(reach_next.sum()),
    "Install label 0 or 0+": int(base["Installs"].isin(["0", "0+"]).sum()),
    "Reviews mean, with exact duplicates": round(base["Reviews_num"].mean(), 1),
    "Reviews mean, exact duplicates removed": round(dedup["Reviews_num"].mean(), 1),
    "Extra rows after exact dedup": int(dedup["App"].duplicated().sum()),
    "Apps that still have extra rows": int(dedup.loc[dedup["App"].duplicated(keep=False), "App"].nunique()),
    "Apps with more than one category": int((dedup.groupby("App")["Category"].nunique() > 1).sum()),
}))
display(base.loc[base["Price_num"] >= 100, ["App", "Category", "Price", "Installs", "Rating"]])
display(base.loc[above_floor, ["App", "Reviews_num", "Installs", "Installs_num", "Install_upper"]])
```

**4.2. Mẫu xuất hiện của dữ liệu thiếu:**  

Nhóm tiếp tục kiểm tra các tổ hợp cột bị thiếu trong cùng một bản ghi và so sánh tỷ lệ thiếu `Rating` giữa các nhóm ứng dụng. Qua đó, nhóm tìm hiểu liệu dữ liệu thiếu có tập trung ở những nhóm cụ thể hay không.

```python
# Mỗi dòng được mô tả bằng danh sách các cột bị thiếu.
missing_patterns = missing_mask.apply(
    lambda row: ", ".join(row.index[row]) if row.any() else "No missing values",
    axis=1
)

pattern_counts = missing_patterns.value_counts()

pattern_summary = pd.DataFrame({
    "Count": pattern_counts,
    "Percent": pattern_counts / len(df_raw) * 100
})

display(pattern_summary.round(2))
```

Vì cột `Rating` thiếu nhiều nhất, nhóm sẽ kiểm tra theo ngưỡng cài đặt (cột `Installs`):

```python
rating_missing_by_installs = (
    df_raw.assign(Rating_missing=missing_mask["Rating"])
    .groupby("Installs", dropna=False)
    .agg(
        Total_records=("Rating_missing", "size"),
        Missing_rating=("Rating_missing", "sum"),
        Missing_percent=("Rating_missing", "mean")
    )
)

rating_missing_by_installs["Missing_percent"] *= 100

display(
    rating_missing_by_installs
    .sort_values("Missing_percent", ascending=False)
    .round(2)
)
```

**Nhận xét:**

- Tỷ lệ thiếu `Rating` có xu hướng giảm khi ngưỡng cài đặt tăng. Cụ thể, nhóm `1+` thiếu 95,52%, nhóm `100+` thiếu 57,02%, nhóm `1,000+` thiếu 21,39%, còn nhóm `1,000,000+` chỉ thiếu 0,13%. Các nhóm có ngưỡng cài đặt từ `5,000,000+` trở lên không ghi nhận giá trị `Rating` bị thiếu.

### **5. Relationships**

**5.1. Tương quan giữa các cột số**

`Reviews` và `Installs` lệch phải, nên Pearson trên thang gốc bị vài ứng dụng cực lớn kéo xuống. Phần này đọc Spearman trên thang gốc và Pearson trên thang log10(x + 1). Các cột số lấy từ `numeric_stats_view`, tức là đã bỏ dòng lệch cột.

```python
corr_cols = ["Rating", "Reviews", "Size", "Installs", "Price"]
display(numeric_stats_view[corr_cols].corr(method="pearson").round(3))
display(numeric_stats_view[corr_cols].corr(method="spearman").round(3))
log_pair = pd.DataFrame({
    "log10(Reviews + 1)": np.log10(numeric_stats_view["Reviews"] + 1),
    "log10(Installs + 1)": np.log10(numeric_stats_view["Installs"] + 1),
})
display(log_pair.corr(method="pearson").round(3))
```

**Nhận xét:**

- Pearson giữa `Reviews` và `Installs` trên thang gốc là 0,64. Spearman là 0,97. Pearson trên log10(x + 1) là 0,96. Hai cột có mối liên hệ mạnh với mức độ phổ biến. Việc có dùng `Reviews` để dự đoán lượt cài hay không phải dựa vào thời điểm dự đoán và thông tin có sẵn; tương quan cao tự nó không đủ để kết luận rò rỉ dữ liệu.
- Pearson giữa `Rating` và `Size` là 0,084, giữa `Rating` và `Price` là −0,022. Spearman của hai cặp này cũng dưới 0,1.

**5.2. Thống kê theo nhóm**

```python
paired = df_raw.loc[numeric_stats_view.index].copy()
paired["Rating_num"] = numeric_stats_view["Rating"]
paired["Installs_num"] = numeric_stats_view["Installs"]
typed = paired[paired["Type"].isin(["Free", "Paid"])].copy()
paid_counts = pd.crosstab(typed["Category"], typed["Type"])
paid_counts["Paid share"] = paid_counts["Paid"] / (paid_counts["Free"] + paid_counts["Paid"])
display(paid_counts.sort_values("Paid share", ascending=False).head(8).round(3))
display(
    typed.groupby("Type").agg(
        rating_count=("Rating_num", "count"),
        rating_mean=("Rating_num", "mean"),
        rating_median=("Rating_num", "median"),
        installs_median=("Installs_num", "median"),
    ).round(3)
)
display(pd.crosstab(typed["Content Rating"], typed["Type"]))
```

**Nhận xét:**

- `MEDICAL` có tỷ lệ trả phí 23,5% (109/463 dòng Free hoặc Paid). `PERSONALIZATION` có 21,2% (83/392). `FAMILY` có nhiều app trả phí nhất về số đếm, 191 dòng, nhưng tỷ lệ chỉ khoảng 9,7%.
- Trung bình `Rating` của Paid là 4,27 và của Free là 4,19. Trung vị là 4,4 và 4,3. Trung vị cận dưới lượt cài của Paid là 1.000, của Free là 100.000.
- `Everyone` là nhãn lớn ở cả Free và Paid. `Adults only 18+` và `Unrated` chỉ có vài dòng, nên không dùng hai nhãn này để chia nhóm giá.

- Có 274 dòng `Rating` bằng 5,0. Trong đó 205 dòng có nhãn cài từ `100+` trở xuống, 178 dòng có tối đa 5 review, và 50 dòng có trên 10 review.
- App trả phí có trung vị `Rating` 4,4 so với 4,3 của app miễn phí, và trung vị cận dưới lượt cài `1.000+` so với `100.000+`. `MEDICAL` và `PERSONALIZATION` có tỷ lệ trả phí cao nhất, khoảng 23,5% và 21,2%.

**2. Lỗi chất lượng dữ liệu nghiêm trọng**

- Một dòng, index 10472, bị lệch cột và có `Rating` bằng 19.
- Có 483 dòng trùng hoàn toàn. Sau khi bỏ chúng còn 698 dòng là bản ghi thêm của 523 ứng dụng trùng tên. 85 ứng dụng có hơn một `Category`.

- `Rating` thiếu 13,6% và tập trung ở app ít lượt cài. `Varies with device` là placeholder của `Size`, 1.695 dòng, không phải giá trị thiếu. Nhóm này có trung vị cận dưới lượt cài `10.000.000+`.

**3. Kế hoạch tiền xử lý**

- Bỏ dòng index 10472.
- Bỏ 483 dòng trùng hoàn toàn, giữ lần xuất hiện đầu tiên.
- Chưa gộp các dòng trùng tên. Gộp theo `Reviews` lớn nhất sẽ làm mất category của 85 ứng dụng có nhiều hơn một category.
- Chuyển `Reviews`, `Price`, `Installs` và `Last Updated` sang kiểu số hoặc ngày. `Installs` sau khi chuyển vẫn là cận dưới.
- Giữ `Size` gốc dưới dạng `string`; tạo `Size_MB` kiểu `float64` theo quy ước giữ giá trị có hậu tố `M` và chia giá trị có hậu tố `k` cho 1.024. Tạo `Size_varies` kiểu `boolean`: những bản ghi `Varies with device` có `Size_MB = NaN` và `Size_varies = True`. Không điền kích thước bằng trung bình; giữ nhãn gốc để so nhóm khi cần.
- Giữ dòng thiếu `Rating`. Không điền trung bình, vì dòng thiếu tập trung ở app ít lượt cài.
- Giữ 20 bản ghi giá từ 100 USD trở lên trong bảng chung. Khi phân tích giá, kiểm tra độ bền bằng cách tạm tách nhóm này và tính lại kết quả.
- Khi gộp bảng review: bỏ dòng không có nội dung review và dòng trùng hoàn toàn, dùng `App` làm khóa, rồi báo số tên không khớp. Nếu yêu cầu review có nội dung, mẫu khớp hiện có 816 tên ứng dụng. Tổng hợp review theo ứng dụng hoặc kiểm soát quan hệ nhiều–nhiều trước khi gộp.

**4. Dấu hiệu có thể dẫn tới câu hỏi**

- Category nào đổi thứ hạng khi điểm 5 sao của app ít review được tính kèm số review bằng điểm trung bình có trọng số (Bayesian rating), và kết quả có đổi khi nâng ngưỡng số review tối thiểu hoặc đổi quy tắc gộp các dòng trùng tên hay không.

- Mức giá và mô hình thu phí có mối liên hệ như thế nào với cận dưới lượt cài đặt giữa các category sau khi kiểm soát nhóm độ tuổi (`Content Rating`)? Kiểm tra lại kết quả khi tạm tách 20 bản ghi giá từ 100 USD trở lên. Không coi tích giá hiện tại và cận dưới lượt cài là doanh thu tối thiểu vì dữ liệu không ghi giá của các giao dịch trong quá khứ hoặc số giao dịch thực tế.

- `Rating` hiển thị trên cửa hàng và cảm xúc bình luận thực tế (`Sentiment_Polarity`) có còn lệch nhau sau khi kiểm soát category và quy mô lượt cài đặt hay không. Mẫu gộp chỉ gồm 816 ứng dụng có nội dung review, không đại diện cho 9.660 tên ứng dụng ở bảng chính.

- Số ngày từ `Last Updated` đến mốc tham chiếu 08/08/2018 có mối liên hệ như thế nào với mức độ phổ biến, sau khi phân nhóm theo category và mô hình miễn phí/trả phí? Mốc này chỉ là ngày cập nhật lớn nhất trong dữ liệu, không phải ngày thu thập đã xác nhận. Dữ liệu không có lịch sử cập nhật nên không dùng resampling/rolling để kết luận xu hướng cập nhật của từng ứng dụng.

- Nhóm ứng dụng có kích thước `Varies with device` có trung vị cận dưới lượt cài đặt cao gấp 100 lần nhóm có kích thước cố định (`10.000.000+` so với `100.000+`). Cần kiểm tra xem khoảng cách này đến từ đặc thù danh mục ứng dụng và yêu cầu phiên bản hệ điều hành (`Android Ver`) hay do cơ chế phân phối gói cài đặt của các ứng dụng lớn.

- Câu hỏi dự đoán ứng dụng đạt ngưỡng lượt cài đặt cao cần xác định rõ thời điểm dự đoán và các đặc trưng có sẵn. Nếu chỉ dùng thông tin mô tả, loại `Reviews` và `Rating` theo phạm vi này; không kết luận leakage chỉ từ tương quan. Dữ liệu hiện là ảnh chụp, không đủ để đánh giá dự đoán thành công trước phát hành. Khi còn nhiều bản ghi cùng ứng dụng, dùng group-based split và group-based cross-validation để giữ cùng ứng dụng ở một phía trong mỗi lần chia.
