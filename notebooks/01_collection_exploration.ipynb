### **2. Numerical columns**

**2.1 Kiểm tra cách biểu diễn và giá trị không hợp lệ**

Nhóm sẽ xem xét năm cột biểu diễn số gồm `Rating`, `Reviews`, `Size`, `Installs` và `Price`. Một số cột như `Size` và `Installs` đang được lưu dưới dạng chuỗi do chứa đơn vị hoặc ký hiệu. Nhóm kiểm tra cách biểu diễn và tạo bảng số tạm để phục vụ thống kê, đồng thời phân biệt giá trị thiếu, giá trị chưa xác định và giá trị không hợp lệ.  

Trước hết, nhóm sẽ hiển thị trước một vài giá trị:

```python
numerical_cols = ["Rating", "Reviews", "Size", "Installs", "Price"]

display(df_raw[numerical_cols].head(5))
```

Trong năm bản ghi đầu, `Rating` được biểu diễn bằng số thập phân, còn `Reviews` và `Price` chứa số nguyên. Cột `Size` chứa hậu tố M, trong khi `Installs` chứa dấu phẩy và dấu +, nên cần chuyển đổi cách biểu diễn trước khi thống kê.

Tiếp theo nhóm sẽ tạo bảng số tạm để chuyển đổi cột `Size` và `Installs`. Với cột `Size`, nhóm dùng quy ước 1.000 KB = 1 MB. Với các giá trị không chuyển được được biểu diễn bằng NaN trong bảng tạm để kiểm tra tiếp.

```python
numeric_view = pd.DataFrame(index=df_raw.index)

numeric_view["Rating"] = pd.to_numeric(
    df_raw["Rating"], errors="coerce"
)

numeric_view["Reviews"] = pd.to_numeric(
    df_raw["Reviews"], errors="coerce"
)

installs_text = (
    df_raw["Installs"].astype("string")
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)
numeric_view["Installs"] = pd.to_numeric(
    installs_text, errors="coerce"
)

price_text = (
    df_raw["Price"].astype("string")
    .str.replace("$", "", regex=False)
)
numeric_view["Price"] = pd.to_numeric(
    price_text, errors="coerce"
)

size_text = df_raw["Size"].astype("string")
size_parts = size_text.str.extract(r"^(\d+(?:\.\d+)?)([Mk])$")

size_number = pd.to_numeric(size_parts[0], errors="coerce")
size_multiplier = size_parts[1].map({"M": 1.0, "k": 0.001})

numeric_view["Size"] = size_number * size_multiplier

numeric_view = numeric_view[numerical_cols]
```

Sau khi tạo bản ghi tạm, nhóm sẽ cho kiểm tra bản ghi này để xem bộ dữ liệu có chứa giá trị thiếu, placeholder và lỗi chuyển đổi hay không

```python
quality_records = []

for col in numerical_cols:
    raw_text = df_raw[col].astype("string")

    original_missing = (
        raw_text.isna()
        | raw_text.str.strip().eq("").fillna(False)
    )

    placeholder = pd.Series(False, index=df_raw.index)
    if col == "Size":
        placeholder = raw_text.eq("Varies with device").fillna(False)

    parse_failed = (
        numeric_view[col].isna()
        & ~original_missing
        & ~placeholder
    )

    quality_records.append({
        "Column": col,
        "Original missing": int(original_missing.sum()),
        "Placeholder": int(placeholder.sum()),
        "Parse failed": int(parse_failed.sum())
    })

    if parse_failed.any():
        print(f"\n{col}: values that could not be parsed")
        display(
            df_raw.loc[parse_failed, ["App", col]]
        )

display(pd.DataFrame(quality_records).set_index("Column"))
```

**Nhận xét**  

- `Rating` có 1.474 giá trị thiếu từ dữ liệu gốc và không phát sinh lỗi chuyển đổi  

- `Size` có 1.695 giá trị Varies with device, thể hiện kích thước chưa xác định cụ 
thể. Đây là placeholder, nhóm sẽ không xếp giá trị này thành giá trị thiếu hay giá trị có lỗi chuyển đổi  

- Các cột `Reviews`, `Size`, `Installs` và `Price` đều có một giá trị không chuyển được theo quy tắc đã đặt, lần lượt là 3.0M, 1,000+, Free và Everyone. Các giá trị này cùng thuộc bản ghi có index 10472, ứng dụng Life Made WI-Fi Touchscreen Photo Frame, cho thấy dấu hiệu lệch cột

Cuối cùng, nhóm sẽ cho kiểm tra tính hợp lệ của các giá trị số trong 5 cột này:  

```python
invalid_masks = {
    "Rating": (
        numeric_view["Rating"].notna()
        & ~numeric_view["Rating"].between(1, 5)
    ),
    "Reviews": (
        numeric_view["Reviews"].lt(0)
        | numeric_view["Reviews"].mod(1).ne(0)
    ) & numeric_view["Reviews"].notna(),
    "Size": numeric_view["Size"].le(0),
    "Installs": (
        numeric_view["Installs"].lt(0)
        | numeric_view["Installs"].mod(1).ne(0)
    ) & numeric_view["Installs"].notna(),
    "Price": numeric_view["Price"].lt(0)
}

for col, mask in invalid_masks.items():
    print(f"\n{col}: {mask.sum()} invalid numeric values")
    display(df_raw.loc[mask, ["App", col]])
```

Sau khi kiểm tra tính hợp lệ của giá trị số, nhóm thấy có một số điểm như sau:  

- Cột `Rating` có một giá trị 19, nằm ngoài thang điểm từ 1 đến 5. Giá trị này cũng thuộc bản ghi bị lệch cột có index 10472.  
- Trong các giá trị chuyển đổi thành công, nhóm không phát hiện `Reviews`, `Installs` hay `Price` chứa giá trị âm hay có phần thập phân. Nhóm cũng không phát hiện `Size` nhỏ hơn hoặc bằng 0.  
- Các giá trị chuyển đổi thất bại ở các cell trước sẽ không được tính là giá trị số hợp lệ. Còn lại kết quả này được nhóm coi là hợp lệ để dùng cho các bước tiếp theo.

**2.2. Thống kê mô tả**

Với bảng dữ liệu tạm, nhóm kiểm tra xem số lượng giá trị hợp lệ, trung bình, trung vị, độ lệch chuẩn, min, max và các tứ phân vị.

```python
# Tạm loại bản ghi lệch cột khỏi phạm vi thống kê.
malformed_rows = (
    df_raw["Category"].eq("1.9")
)

numeric_stats_view = numeric_view.loc[~malformed_rows].copy()

# Loại các giá trị vi phạm điều kiện đã kiểm tra ở 2.1
# khỏi thống kê tạm, không thay đổi df_raw hoặc numeric_view.
for col in numerical_cols:
    numeric_stats_view[col] = numeric_stats_view[col].mask(
        invalid_masks[col].loc[numeric_stats_view.index]
    )

stats_summary = (
    numeric_stats_view.describe()
    .T
    .rename(columns={
        "count": "Valid count",
        "mean": "Mean",
        "std": "Std",
        "min": "Min",
        "25%": "Q1",
        "50%": "Median",
        "75%": "Q3",
        "max": "Max"
    })
)

stats_summary["IQR"] = stats_summary["Q3"] - stats_summary["Q1"]
stats_summary["Range"] = stats_summary["Max"] - stats_summary["Min"]

with pd.option_context("display.float_format", "{:,.3f}".format):
    display(stats_summary)
```

**2.4. Giá trị ngoại lệ và giá trị bất thường**  

Ở bước cuối nhóm sẽ tìm các điểm cực trị, đối chiếu bản ghi và phân biệt lỗi dữ liệu với giá trị lớn nhưng hợp lệ.

```python
outlier_masks = {}
outlier_records = []

for col in numerical_cols:
    values = numeric_stats_view[col]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask = values.notna() & (
        values.lt(lower) | values.gt(upper)
    )
    outlier_masks[col] = mask

    valid_count = values.notna().sum()

    outlier_records.append({
        "Column": col,
        "Lower bound": lower,
        "Upper bound": upper,
        "IQR": iqr,
        "Flagged count": int(mask.sum()),
        "Flagged percent": (
            mask.sum() / valid_count * 100
            if valid_count else float("nan")
        )
    })

outlier_summary = pd.DataFrame(outlier_records).set_index("Column")
display(outlier_summary.round(3))
```

```python
for col in numerical_cols:
    flagged_values = numeric_stats_view.loc[outlier_masks[col], col]

    selected_index = (
        flagged_values.nsmallest(5).index
        .union(flagged_values.nlargest(5).index, sort=False)
    )

    examples = df_raw.loc[
        selected_index,
        ["App", "Category", col]
    ].copy()

    examples["Numeric value"] = numeric_stats_view.loc[
        selected_index, col
    ]

    print(f"\n{col}: examples flagged by the IQR rule")
    display(examples.sort_values("Numeric value"))
```

```python
# Mỗi dòng được mô tả bằng danh sách các cột bị thiếu.
missing_patterns = missing_mask.apply(
    lambda row: ", ".join(row.index[row]) if row.any() else "No missing values",
    axis=1
)

pattern_counts = missing_patterns.value_counts()

pattern_summary = pd.DataFrame({
    "Count": pattern_counts,
    "Percent": pattern_counts / len(df_raw) * 100
})

display(pattern_summary.round(2))
```

Vì cột `Rating` thiếu nhiều nhất, nhóm sẽ kiểm tra theo ngưỡng cài đặt (cột `Installs`):

```python
rating_missing_by_installs = (
    df_raw.assign(Rating_missing=missing_mask["Rating"])
    .groupby("Installs", dropna=False)
    .agg(
        Total_records=("Rating_missing", "size"),
        Missing_rating=("Rating_missing", "sum"),
        Missing_percent=("Rating_missing", "mean")
    )
)

rating_missing_by_installs["Missing_percent"] *= 100

display(
    rating_missing_by_installs
    .sort_values("Missing_percent", ascending=False)
    .round(2)
)
```

- Có 9.360 bản ghi (86,34%) không thiếu giá trị và 1.481 bản ghi (13,66%) thiếu ít nhất một thuộc tính. Trong đó, 1.469 bản ghi chỉ thiếu `Rating`, cho thấy dữ liệu thiếu chủ yếu tập trung ở cột này.  

- Các trường hợp thiếu nhiều thuộc tính cùng lúc có số lượng nhỏ: 4 bản ghi thiếu cả `Rating` và `Current Ver`, 1 bản ghi thiếu cả `Rating` và `Type`, và 1 bản ghi thiếu cả `Content Rating` và `Android Ver`. Trường hợp cuối thuộc bản ghi có dấu hiệu lệch cột đã được xác định ở mục 3.3, nên cần phân biệt với các trường hợp thiếu thông thường.  

- Tỷ lệ thiếu `Rating` có xu hướng giảm khi ngưỡng cài đặt tăng. Cụ thể, nhóm `1+` thiếu 95,52%, nhóm `100+` thiếu 57,02%, nhóm `1,000+` thiếu 21,39%, còn nhóm `1,000,000+` chỉ thiếu 0,13%. Các nhóm có ngưỡng cài đặt từ `5,000,000+` trở lên không ghi nhận giá trị `Rating` bị thiếu.  

- Như vậy, việc thiếu `Rating` tập trung nhiều ở các bản ghi có ngưỡng số lượt cài đặt thấp. Đây là mối liên hệ nhóm quan sát được trong dataset, tuy vậy vẫn chưa đủ thông tin để khẳng định nguyên nhân hay cơ chế gây thiếu dữ liệu.

**4.3. Nguyên nhân có thể và hướng xử lý dự kiến:**  

Dựa trên kết quả thống kê và mẫu xuất hiện của dữ liệu thiếu, nhóm đề xuất hướng xử lý cho từng cột như sau. Các nguyên nhân dưới đây là giả thuyết từ dữ liệu quan sát được, chưa được xác nhận từ nguồn thu thập.  
| Cột | Nguyên nhân thiếu dữ liệu | Hướng xử lý dự kiến |  
|---|---|---|  
| `Rating` | Ứng dụng có thể chưa có đủ người dùng để đánh giá hoặc thông tin không được thu thập đầy đủ | Giữ các bản ghi cho những phân tích không cần `Rating`. Khi phân tích điểm đánh giá, nhóm sẽ cân nhắc sử dụng phương pháp đánh giá điểm theo trọng số |  
| `Current Ver` | Thông tin phiên bản có thể không có trên nguồn hoặc bị bỏ sót khi thu thập | Nhóm có thể điền giá trị thiếu bằng nhãn "Unknown" nếu giữ cột |  
| `Android Ver` | Một trường hợp thuộc bản ghi lệch cột. Các trường hợp còn lại có thể do thông tin không được cung cấp hoặc thu thập đầy đủ | Nhóm sẽ xem xét loại bỏ bản ghi lệch cột ở bước sau. Với các trường hợp còn lại, nhóm có thể điền giá trị thiếu bằng nhãn "Unknown" nếu giữ cột |  
| `Type` | Chưa xác định được nguyên nhân từ kết quả hiện tại | Kiểm tra bản ghi thiếu và đối chiếu với `Price`. Nếu giá hợp lệ, nhóm sẽ cân nhắc suy ra "Free" khi giá bằng 0 hoặc "Paid" khi giá lớn hơn 0 |   
| `Content Rating` | Giá trị thiếu nằm trong bản ghi có dấu hiệu lệch cột | Xem xét loại toàn bộ bản ghi lệch cột ở bước sau |
